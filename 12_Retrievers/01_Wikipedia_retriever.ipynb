## Wikipedia Retriever 
Wikipedia Retriver is a retriever that queries the Wikipedia API to fetch relevant document

```python
# !pip install langchain chromadb faiss-cpu langchain_huggingface langchain-community wikipedia
```

```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
# Initialize the retiever (optional: specify language and top_k)
retriever = WikipediaRetriever(language="en", top_k=5)
```

```python
# Define your query
query = "Attention All You Need"

# Get relevant dcouments
docs = retriever.invoke(query)
```

```python
# Print the retrieved documents
for i, doc in enumerate(docs):
    print(f"Document {i+1}:")
    print(f"Title: {doc.metadata.get('title', 'N/A')}")
    print(f"Content: {doc.page_content[:500]}...")  # Print first 500 characters
    # print(f"Content: {doc.page_content}") # Print full article
    print("-" * 80)
```
